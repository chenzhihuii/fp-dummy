# Cleaning, Encode, Normalisasi

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler


print(df.head(10).to_string(index=False))

print("\n=== Cleaning Data ===")
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

print(f"Jumlah data setelah cleaning: {df.shape[0]} baris")

# === 3. Pisahkan fitur numerik dan kategorikal ===
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\nFitur Numerik:", num_cols)
print("Fitur Kategorikal:", cat_cols)

# === 4. Encoding kategorikal (One-Hot Encoding) ===
df_encoded = pd.get_dummies(df, columns=cat_cols)

# === 5. Normalisasi Numerik (MinMaxScaler) ===
scaler = MinMaxScaler()
df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])

print("\n=== Data Setelah Encoding & Normalisasi (10 Baris Pertama) ===")
print(df_encoded.head(10).to_string(index=False))

output_file = 'data_clean.csv'
df_encoded.to_csv(output_file, index=False)
print(f"\nData berhasil disimpan ke: {output_file}")
```

# Cek perlu PCA ato engga

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

numeric_cols = df.select_dtypes(include=[np.number]).columns
data_numeric = df[numeric_cols]

# korelasi
corr_matrix = data_numeric.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Heatmap Korelasi")
plt.show()

# Cek korelasi kuat
high_corr = (abs(corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))) > 0.8).sum().sum()

if high_corr > 0:
    print(f"Ada {high_corr} pasang variabel dengan korelasi > 0.8. Disarankan melakukan PCA.")
    
    # Standardisasi
    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(data_numeric)
    
    # 5. PCA (2 komponen)
    pca = PCA(n_components=2)
    pca_result = pca.fit_transform(data_scaled)
    
    print("Explained Variance Ratio:", pca.explained_variance_ratio_)
    
    # df hasil PCA
    df_pca = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])

    plt.figure(figsize=(8,6))
    plt.scatter(df_pca['PC1'], df_pca['PC2'], alpha=0.7, color='teal')
    plt.title('Visualisasi PCA (2 Komponen Utama)')
    plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}%)")
    plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}%)")
    plt.grid(True)
    plt.show()
else:
    print("Tidak ada korelasi kuat antarvariabel, PCA tidak diperlukan.")
```

# Kmeans klaster

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

data = pd.read_csv("D:/Semester 7 (EVAL PA)/Try Code/data_clean.csv")
numerical_data = data.select_dtypes(include=[np.number])

# Standardisasi
scaler = StandardScaler()
X = scaler.fit_transform(numerical_data)


inertia = []
K_range = range(2, 11) #ini masih make sense ga? please cek
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(6,4))
plt.plot(K_range, inertia, 'bo-', linewidth=2)
plt.xlabel("Jumlah Klaster (k)")
plt.ylabel("Inertia / Sum of Squared Distances")
plt.title("Elbow Method untuk Menentukan k Optimal")
plt.show()

sil_scores = []
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X)
    sil = silhouette_score(X, labels)
    sil_scores.append(sil)

plt.figure(figsize=(6,4))
plt.plot(K_range, sil_scores, 'ro-', linewidth=2)
plt.xlabel("Jumlah Klaster (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Analysis untuk Menentukan k Optimal")
plt.show()

k_optimal = K_range[np.argmax(sil_scores)]
print(f"Jumlah klaster optimal berdasarkan Silhouette Score: {k_optimal}")

kmeans_final = KMeans(n_clusters=k_optimal, random_state=42)
labels_final = kmeans_final.fit_predict(X)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

data = pd.read_csv("D:/Semester 7 (EVAL PA)/Try Code/data_clean.csv")
numerical_data = data.select_dtypes(include=[np.number])

scaler = StandardScaler()
X = scaler.fit_transform(numerical_data)

# 
k = 5
kmeans = KMeans(n_clusters=k, random_state=42)
labels_kmeans = kmeans.fit_predict(X)

# Evaluasi performa
sil_k = silhouette_score(X, labels_kmeans)
dbi_k = davies_bouldin_score(X, labels_kmeans)

# Distribusi anggota per klaster
cluster_counts = pd.Series(labels_kmeans).value_counts()

print("=== K-Means Clustering ===")
print(f"Silhouette Score: {sil_k}")
print(f"Davies-Bouldin Index: {dbi_k}")
print("Distribusi anggota per klaster:")
print(cluster_counts)
```

# OPTICS

```python
# optics_plot.py
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import OPTICS
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, davies_bouldin_score

data = pd.read_csv("D:/Semester 7 (EVAL PA)/Try Code/data_clean.csv")
numerical_data = data.select_dtypes(include=[np.number])
scaler = StandardScaler()
X = scaler.fit_transform(numerical_data)

# OPTICS
optics = OPTICS(min_samples=5, xi=0.02, min_cluster_size=0.05)
labels_optics = optics.fit_predict(X)
unique_labels = set(labels_optics)
if len(unique_labels - {-1}) <= 1:
    optics = OPTICS(min_samples=5, xi=0.01, min_cluster_size=0.03)
    labels_optics = optics.fit_predict(X)

# Evaluasi (buang noise)
mask = labels_optics != -1
if len(set(labels_optics[mask])) > 1:
    sil_o = silhouette_score(X[mask], labels_optics[mask])
    dbi_o = davies_bouldin_score(X[mask], labels_optics[mask])
else:
    sil_o = None
    dbi_o = None

print("=== OPTICS Clustering ===")
print(f"Silhouette Score: {sil_o}")
print(f"Davies-Bouldin Index: {dbi_o}")
print(pd.Series(labels_optics).value_counts())

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(8,6))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=labels_optics, palette='tab10', s=60)
plt.title("OPTICS Clustering - PCA 2D")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()
```
